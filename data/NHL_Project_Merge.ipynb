# Scrape API 
---

```python
#format date
project_df["game_date"] = project_df["date_time_CT"].dt.strftime("%Y-%m-%d")
```

```python
#drop duplicate dates and show unique dates (and count of unique dates)
unique_dates = sorted(project_df["game_date"].dropna().unique())
print(len(unique_dates))
#print(unique_dates)
```

```python
#function that gets and scrapes API
def get_nhl_json(url):
    response = requests.get(url, timeout=30)
    response.raise_for_status()
    return response.json()

#empty list
all_api_games = []

for date_str in unique_dates:
    url = f"https://api-web.nhle.com/v1/score/{date_str}" #url
    
    try:
        data = get_nhl_json(url)
        
        if "games" in data and len(data["games"]) > 0:
            day_df = json_normalize(data["games"])
            day_df["query_date"] = date_str
            all_api_games.append(day_df)
            
    except Exception as e:
        print(f"Error on {date_str}: {e}")
```

```python
#concat list to dataframe then display shape and head
api_games_df = pd.concat(all_api_games, ignore_index=True)
print(api_games_df.shape)
api_games_df.head(5)
```

```python
#drop unecessary columns
api_games_df = api_games_df[[
    "id",
    "season",
    "gameDate",
    "gameType",
    "gameState",
    "awayTeam.id",
    "awayTeam.abbrev",
    "awayTeam.score",
    "homeTeam.id",
    "homeTeam.abbrev",
    "homeTeam.score",
    "venue.default",
    "seriesStatus.seriesTitle",
    "seriesStatus.round"
]].copy()
```

```python
#rename columns
api_games_df.columns = [
    "game_id",
    "season_api",
    "game_date_api",
    "game_type_api",
    "game_state_api",
    "away_team_id_api",
    "away_team_abbrev_api",
    "away_score_api",
    "home_team_id_api",
    "home_team_abbrev_api",
    "home_score_api",
    "venue_api", 
    "playoff_title_api", 
    "playoff_round_api"
]
```
